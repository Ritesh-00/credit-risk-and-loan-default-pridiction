```python
from sklearn.base import BaseEstimator, TransformerMixin

class FeatureEngineer(BaseEstimator, TransformerMixin):

    def __init__(self, cols_to_drop=None,categorical_numeric=None):
        self.cols_to_drop = cols_to_drop
        self.categorical_numeric = categorical_numeric

    def fit(self, X, y=None):
        return self



    def transform(self, X):

        X = X.copy()
        # Missingness
        X['TOTAL_MISSING'] = X.isna().sum(axis=1)
        X['MISSING_RATIO'] = X.isna().mean(axis=1)

        for col in self.categorical_numeric:
            if col in X.columns:

                X[col] = X[col].apply(
                    lambda value: (
                        f'UNKNOWN_{col}' if pd.isna(value)
                        else f'{col}_{value}'
                    )
                )

        # Age
        X['AGE_YEARS'] = (
            -X['DAYS_BIRTH'] / 365.25
        )

        # Employment
        X['EMPLOYED_ANOMALY'] = (
            X['DAYS_EMPLOYED'] > 0
        ).astype(int)

        X['EMPLOYED_YEARS'] = (
            X['DAYS_EMPLOYED']
            .where(X['DAYS_EMPLOYED'] <= 0)
            / -365.25
        )

        # Financial ratios
        X['CREDIT_INCOME_RATIO'] = (
            X['AMT_CREDIT'] /
            X['AMT_INCOME_TOTAL'].replace(0, np.nan)
        )

        X['ANNUITY_INCOME_RATIO'] = (
            X['AMT_ANNUITY'] /
            X['AMT_INCOME_TOTAL'].replace(0, np.nan)
        )

        X['GOODS_INCOME_RATIO'] = (
            X['AMT_GOODS_PRICE'] /
            X['AMT_INCOME_TOTAL'].replace(0, np.nan)
        )

        # Family
        X['INCOME_PER_FAMILY_MEMBER'] = (
            X['AMT_INCOME_TOTAL'] /
            X['CNT_FAM_MEMBERS'].replace(0, np.nan)
        )

        X['OWN_CAR_AGE_MISSING'] = (
            X['OWN_CAR_AGE'].isna()
        ).astype(int)

        if 'OWN_CAR_AGE' in X.columns:
            X['OWN_CAR_AGE'] = X['OWN_CAR_AGE'].fillna(0)

        # Drop selected columns
        X = X.drop(
            columns=self.cols_to_drop,
            errors='ignore'
        )

        return X
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

numeric_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median', add_indicator=True)),
    ('scaler', StandardScaler())
])

categorical_pipeline_org = Pipeline([
    ('imputer', SimpleImputer(strategy='constant', fill_value='Unknown' )),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])


preprocessor = ColumnTransformer([
    ('num', numeric_pipeline, numeric_features),

    ('cat_org', categorical_pipeline_org, categorical_features),
])


```

```python
pipelines = {}

for name,model in models.items():
  pipelines[name] = Pipeline([
      ('feature_engineering', FeatureEngineer(
          cols_to_drop=coltodrop,
          categorical_numeric=categorical_numeric
        )),
      ('preprocessor', preprocessor),
      ('model', model)
  ])
```

```python
class CatBoostPreprocessor(BaseEstimator, TransformerMixin):

    def __init__(self, categorical_features=None):
        self.categorical_features = categorical_features

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        X = X.copy()

        for col in self.categorical_features:
            if col in X.columns:
                X[col] = (
                        X[col]
                        .fillna('Unknown')
                        .astype(str)
                )
        return X


```
